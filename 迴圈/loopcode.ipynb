# Loop
Loop 指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。

```python
# 底下的程式碼會印出 0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
# 底下的程式碼會印出 0-9
for i in range(10):
    print(i+1)
```

# Range
Python:
- 懶惰是一種美德
- 由0開頭
- 包含頭，不包含尾

```python
print(list(range(5)))
#區間，由0開始，到5-1結束
print(list(range(0,5)))
#區間，由1開始，到5-1結束
print(list(range(1,5)))
#區間，由1開始，到5-1結束，每次跳兩個
print(list(range(1,5,2)))
```

```python
print(list(range(2,101,2)))
```

# 九九乘法表
1\*1=1﹐1\*2=2，1\*3=3，....  
2\*1=2﹐2\*2=4，2\*3=6，....  
3\*1=3﹐3\*2=6，3\*3=9，....  

```python
i=1
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end=',')
print()
i=2
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end=',')
print()
i=3
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end=',')
print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=' ')
    print()
```

# 練習:印出質數
質數:一個數字$n$，除了自己本身，以及1以外，沒有任何數可以除盡他，n就是質數。

```python
n=273
i=2

if 273 % i== 0:
    print(f'{273}不是質數')

i=3
if 273 % i== 0:
    print(f'{273}不是質數')

i=4
if 273 % i== 0:
    print(f'{273}不是質數')
```

```python
n=122
isPrime = True # 假設 n 是質數
for i in range(2,n):
    if n % i== 0:
        print(f'{n}不是質數，因為它可以被{i}除盡')
        isPrime = False # 確定 n 不是質數
        break # 只要數字能被除盡，就跳離迴圈

if isPrime == True: # 判斷我們的假設是否還成立
    print(f'{n}是質數')
    
```
