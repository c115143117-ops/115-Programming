# 介紹
使用 pyautogui 來完成小恐龍的遊戲機器人。

小恐龍是在 chrom 中使用 chrome://dino 啟動。

邏輯:
- 不斷監視畫面
- 讀取某個位置的顏色
- 若此位置上出現(83,83,83)的顏色
- 跳起來

```python
import pyautogui as pyg

pyg.sleep(2)        # sleep 2 s
pyg.press('space')  # 啟動遊戲

jump_count = 0      # 記錄跳躍次數（用來估算分數）

# 無窮迴圈
while True:
    im = pyg.screenshot()
    bg_color = im.getpixel((676, 193))  # 動態抓取背景顏色
    obs = False

    # -------------------------------------------------------------
    # 動態調整參數：根據跳躍次數（代表遊戲進行時間與分數）調整延遲
    # 跳躍大約超過 40~50 次（相當於 900 分以上高速狀態）
    # -------------------------------------------------------------
    if jump_count >= 40:
        air_time = 0.03     # 跳躍滯空時間縮短（原本 0.14 -> 0.03）
        duck_time = 0.1     # 下蹲維持時間縮短（原本 0.29 -> 0.1）
        detect_range = 30   # 延伸偵測範圍（15 -> 30 像素），提早發現障礙物
    else:
        air_time = 0.14     # 前期正常速度 
        duck_time = 0.29
        detect_range = 15 
 
    for i in range(detect_range): 
        c_cactus = im.getpixel((230 + i, 375))  # 仙人掌
        c_bird = im.getpixel((230 + i, 330))    # 翼龍
       
        if c_cactus != bg_color: #仙人掌，比對背景顏色
            pyg.keyDown('space')
            pyg.keyUp('space')            
            pyg.sleep(air_time) # 使用動態滯空時間（900分後變超短）            
            pyg.keyDown('down') # 強制快速下蹲落地
            pyg.sleep(0.00001) # 保持下蹲姿勢 0.00001 秒
            pyg.keyUp('down')
            
            obs = True
            jump_count += 1 # 累加跳躍次數
            break

        if c_bird != bg_color: # 翼龍，比對背景顏色
            pyg.keyDown('down')
            pyg.sleep(duck_time) # 使用動態下蹲時間（900分後快速站起）
            pyg.keyUp('down')
            
            obs = True
            jump_count += 1  # 累加跳躍次數
            break

    if not obs:
        pass
```
